# Pruebas propias

Estos son casos que hice aparte para revisar situaciones diferentes

```python
def ordenar_insertion(precios, opcion):
    precios = precios.copy()

    for i in range(1, len(precios)):
        clave = precios[i]
        j = i - 1

        while j >= 0 and precios[j] > clave:
            precios[j + 1] = precios[j]
            j = j - 1

        precios[j + 1] = clave

    if opcion == "2":
        precios.reverse()

    return precios
```

## Caso 1 - Precios repetidos

```python
precios = [5000, 2000, 5000, 1000, 2000, 3000]
esperada = [1000, 2000, 2000, 3000, 5000, 5000]

resultado = ordenar_insertion(precios, "1")
print("Salida obtenida:", resultado)
print("Salida esperada:", esperada)

assert resultado == esperada
print("Prueba pasada")
```

## Caso 2 - Lista ya ordenada

```python
precios = [1000, 2000, 3000, 4000, 5000]
esperada = [1000, 2000, 3000, 4000, 5000]

resultado = ordenar_insertion(precios, "1")
print("Salida obtenida:", resultado)
print("Salida esperada:", esperada)

assert resultado == esperada
print("Prueba pasada")
```

## Caso 3 - Orden descendente

```python
precios = [1000, 2000, 3000, 4000, 5000]
esperada = [5000, 4000, 3000, 2000, 1000]

resultado = ordenar_insertion(precios, "2")
print("Salida obtenida:", resultado)
print("Salida esperada:", esperada)

assert resultado == esperada
print("Prueba pasada")
```
