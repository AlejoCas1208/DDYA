# Pruebas del enunciado

Probe las dos opciones que pide el problema

```python
def ordenar_insertion(precios, opcion):
    precios = precios.copy()

    for i in range(1, len(precios)):
        clave = precios[i]
        j = i - 1

        while j >= 0 and precios[j] > clave:
            precios[j + 1] = precios[j]
            j = j - 1

        precios[j + 1] = clave

    if opcion == "2":
        precios.reverse()

    return precios
```

## Caso 1 - Ascendente

Precios: 2500, 1000, 5000, 3000, 1500. Opcion: 1

Salida esperada: 1000, 1500, 2500, 3000, 5000

```python
precios = [2500, 1000, 5000, 3000, 1500]
esperada = [1000, 1500, 2500, 3000, 5000]

resultado = ordenar_insertion(precios, "1")
print("Salida obtenida:", resultado)
print("Salida esperada:", esperada)

assert resultado == esperada
print("Prueba pasada")
```

## Caso 2 - Descendente

Precios: 2500, 1000, 5000, 3000, 1500. Opción: 2

Salida esperada: 5000, 3000, 2500, 1500, 1000

```python
precios = [2500, 1000, 5000, 3000, 1500]
esperada = [5000, 3000, 2500, 1500, 1000]

resultado = ordenar_insertion(precios, "2")
print("Salida obtenida:", resultado)
print("Salida esperada:", esperada)

assert resultado == esperada
print("Prueba pasada")
```
