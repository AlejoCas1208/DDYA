### 5) Análisis de complejidad **[15 pts]**
Incluye:






- JUSTIFICACION : 
    - Sin Memoizacion : cada llamada genera genera dos ramificaciones, se crearia un arbol enorme de llamadas repetidas. Eso hace que el tiempo aumente exponencialmete porque basicamente 
    entre mas grande sea n, se vuelve mas dificil calcularlo sin ayuda, da igual mejor y peor caso, porque siempre dependera de n 

    - Con Memoizacion : como cada valor solo se calcula una vez y de ahi en adelante solo se consulta, el tiempo baja, o sea a forma lineal. Esto reduce la complejidad en tiempo ( por el arreglo de memoizacion)

- FUNCION T(N) : 
    - Sin memoizacion (peor caso) : T(n) = T(n-1) + T(n-2) + 1, con T(0) = 1 y T(1) = 1 - Esto crece de forma exponencial, porque cada llamada genera dos llamdas mas sin reutilizar nada
    - Con memoizacion (mejor caso) : T(n) = O(n)  - Prique cada subproblema de O a n se calcula solo una vez 







## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
# Escribe tu código aquí.

def calcularPlan(n, memo):
    if n == 0:
        return 0
    if n == 1:
        return 1
    if memo[n] != -1:
        return memo[n]
    resultado = calcularPlan(n-1, memo) + calcularPlan(n-2, memo)
    memo[n] = resultado
    return resultado

def main():
    n = int(input("Ingrese el mes n : "))
    memo = [-1] * (n + 1)
    resultado = calcularPlan (n, memo)
    print("En el mes",n , "se producieron", resultado, "nuevas plantas")


main()

```
