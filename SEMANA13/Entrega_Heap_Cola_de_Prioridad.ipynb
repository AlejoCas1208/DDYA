## 1) Descripción del problema **[1 pts]**

Un centro de cómputo recibe tareas que deben ser procesadas por sus servidores. Cada tarea tiene un **código**, un **nombre** y una **prioridad** numérica; las tareas con mayor prioridad deben procesarse primero (por ejemplo, `T03 - Renderizar video - Prioridad 10` antes que `T04 - Procesar correos - Prioridad 3`).

Para administrarlas se usa un **Max-Heap implementado sobre un arreglo** (sin nodos ni punteros), de modo que siempre se pueda consultar o extraer en poco tiempo la tarea más urgente. El programa debe:

1. Registrar `n` tareas ingresadas por el usuario, conservando el orden de entrada.
2. Construir el Max-Heap con `MAX-HEAPIFY` y `BUILD-MAX-HEAP`, mostrando el arreglo y una representación del árbol.
3. Ofrecer una cola de prioridad: `MAXIMUM`, `EXTRACT-MAX` e inserción de nuevas tareas.
4. Ordenar las prioridades con `HEAPSORT` implementado manualmente (sin `sort()`, `sorted()` ni `heapq`).
5. Probarse con al menos 6 tareas, evidenciando cada operación.

### 5) Análisis de complejidad **[15 pts]**

**Justificación**

El heap es un árbol binario casi completo guardado en un arreglo: con índices desde 0, el padre de `i` es `(i-1)//2`, el hijo izquierdo `2i+1` y el derecho `2i+2`. Su altura es `h = ⌊log₂ n⌋`, y de ahí salen todos los costos.

| Operación | Justificación | Complejidad |
|---|---|---|
| `MAX-HEAPIFY` | En cada nivel hace como máximo 2 comparaciones y un intercambio; baja a lo sumo `h` niveles | O(log n) |
| `BUILD-MAX-HEAP` | Llama a `MAX-HEAPIFY` en los nodos internos; un nodo a altura `k` cuesta O(k) y hay a lo sumo ⌈n/2^(k+1)⌉ nodos a esa altura: Σ (n/2^(k+1))·k = O(n) | O(n) |
| `MAXIMUM` | Solo lee `A[0]` | O(1) |
| `EXTRACT-MAX` | Mueve el último a la raíz y hace un `MAX-HEAPIFY` | O(log n) |
| Inserción | Agrega al final y sube como máximo `h` niveles | O(log n) |
| `HEAPSORT` | `BUILD-MAX-HEAP` + (n-1) llamadas a `MAX-HEAPIFY` | O(n log n) |

**Función T(n) de HeapSort** (contando comparaciones, con 2 por nivel en `MAX-HEAPIFY`):

- Construcción: T_build(n) ≤ 2n
- Fase de ordenamiento: T_orden(n) ≤ Σ_{k=1}^{n-1} 2·⌊log₂ k⌋ ≤ 2(n-1)·log₂ n
- **T(n) ≤ 2n + 2(n-1)·log₂ n  =  O(n log n)**

**Mejor y peor caso**

- **Peor caso:** O(n log n). Cada extracción hunde la raíz hasta el fondo (por ejemplo, con una lista ascendente `[0, 1, 2, ..., n-1]`). T(n) ≈ 2n·log₂ n.
- **Mejor caso:** con todos los elementos iguales, `MAX-HEAPIFY` nunca intercambia y termina en O(1) en cada llamada, así que T(n) ≈ 3n = O(n). Para elementos distintos, HeapSort sigue siendo Θ(n log n).
- **Espacio:** O(1) adicional (HeapSort in-place sobre el arreglo).

La gráfica del mejor y peor caso (comparaciones reales contadas vs. n) se genera al ejecutar el `main()` de la sección 6.

```python

CONT = {"comp": 0}


def clave(x):
    """Prioridad de un elemento: una tarea (codigo, nombre, prioridad) o un número."""
    return x[2] if isinstance(x, (tuple, list)) else x


def padre(i):
    return (i - 1) // 2


def izq(i):
    return 2 * i + 1


def der(i):
    return 2 * i + 2


def texto_tarea(t):
    return f"{t[0]} - {t[1]}"


def prioridades(A):
    return [clave(t) for t in A]



def max_heapify(A, i, tam):
    """Hunde A[i] hasta que se cumpla la propiedad de Max-Heap en el subárbol."""
    while True:
        l, r = izq(i), der(i)
        mayor = i
        if l < tam:
            CONT["comp"] += 1
            if clave(A[l]) > clave(A[mayor]):
                mayor = l
        if r < tam:
            CONT["comp"] += 1
            if clave(A[r]) > clave(A[mayor]):
                mayor = r
        if mayor == i:
            return
        A[i], A[mayor] = A[mayor], A[i]
        i = mayor


def build_max_heap(A):
    """Convierte el arreglo A en un Max-Heap (de abajo hacia arriba)."""
    for i in range(len(A) // 2 - 1, -1, -1):
        max_heapify(A, i, len(A))


def es_max_heap(A):
    """Verifica: prioridad del padre >= prioridad de sus hijos."""
    for i in range(1, len(A)):
        if clave(A[padre(i)]) < clave(A[i]):
            return False
    return True



def maximum(A):
    """Devuelve la tarea de la raíz sin eliminarla."""
    if len(A) == 0:
        return None
    return A[0]


def extract_max(A):
    """Retira y devuelve la tarea de mayor prioridad, reorganizando el heap."""
    if len(A) == 0:
        return None
    maximo = A[0]
    ultimo = A.pop()
    if len(A) > 0:
        A[0] = ultimo
        max_heapify(A, 0, len(A))
    return maximo


def insertar(A, tarea):
    """Inserta una tarea y la sube hasta conservar la propiedad. Devuelve su posición."""
    A.append(tarea)
    i = len(A) - 1
    while i > 0 and clave(A[padre(i)]) < clave(A[i]):
        A[i], A[padre(i)] = A[padre(i)], A[i]
        i = padre(i)
    return i


def heapsort(lista):
    """Ordena de menor a mayor sin sort()/sorted(). Devuelve (max_heap, ordenada)."""
    A = list(lista)
    build_max_heap(A)
    heap = A[:]
    for fin in range(len(A) - 1, 0, -1):
        A[0], A[fin] = A[fin], A[0]
        max_heapify(A, 0, fin)
    return heap, A



def imprimir_arbol(A):
    """Dibuja el heap (solo prioridades) por niveles, en texto."""
    n = len(A)
    if n == 0:
        print("(heap vacío)")
        return
    P = prioridades(A)
    niveles = n.bit_length()
    ancho = 4 * 2 ** (niveles - 1)
    for nivel in range(niveles):
        ini = 2 ** nivel - 1
        fin = min(2 * ini + 1, n)
        sep = ancho // (2 ** nivel)
        print("".join(str(x).center(sep) for x in P[ini:fin]).rstrip())
        if nivel < niveles - 1:
            linea = ""
            for i in range(ini, fin):
                c = ("/" if izq(i) < n else " ") + " " + ("\\" if der(i) < n else " ")
                linea += c.center(sep)
            print(linea.rstrip())



def leer_entero(mensaje, minimo=None):
    while True:
        try:
            v = int(input(mensaje))
            if minimo is not None and v < minimo:
                print(f"  Debe ser un entero >= {minimo}.")
                continue
            return v
        except ValueError:
            print("  Ingresa un número entero válido.")


def leer_tarea(codigos_usados):
    while True:
        codigo = input("  Código: ").strip()
        if codigo == "":
            print("  El código no puede estar vacío.")
        elif codigo in codigos_usados:
            print("  Ese código ya existe, usa otro.")
        else:
            break
    nombre = input("  Nombre: ").strip()
    prioridad = leer_entero("  Prioridad: ")
    return (codigo, nombre, prioridad)


def registrar_tareas():
    n = leer_entero("Cantidad de tareas (n): ", minimo=1)
    tareas, usados = [], set()
    for k in range(1, n + 1):
        print(f"Tarea {k}:")
        t = leer_tarea(usados)
        usados.add(t[0])
        tareas.append(t)
    return tareas



def contar_heapsort(lista):
    CONT["comp"] = 0
    heapsort(lista)
    return CONT["comp"]


def graficar_complejidad():
    """Mejor caso (todo igual) vs peor caso (ascendente) de HeapSort."""
    import math
    import matplotlib.pyplot as plt
    tamanos = list(range(50, 1001, 50))
    mejor = [contar_heapsort([7] * n) for n in tamanos]
    peor = [contar_heapsort(list(range(n))) for n in tamanos]
    ref = [2 * n * math.log2(n) for n in tamanos]
    print(f"{'n':>6} {'mejor caso':>12} {'peor caso':>12}")
    for n, m, p in zip(tamanos, mejor, peor):
        if n in (50, 250, 500, 1000):
            print(f"{n:>6} {m:>12} {p:>12}")
    plt.figure(figsize=(8, 5))
    plt.plot(tamanos, mejor, "o-", label="Mejor caso (elementos iguales) ~ O(n)")
    plt.plot(tamanos, peor, "s-", label="Peor caso (lista ascendente) ~ O(n log n)")
    plt.plot(tamanos, ref, "--", label="Referencia 2·n·log2(n)")
    plt.xlabel("n (número de tareas)")
    plt.ylabel("Comparaciones")
    plt.title("Complejidad de HeapSort")
    plt.legend()
    plt.grid(True)
    plt.show()



def pruebas():
    import random
    # Ejemplo del enunciado
    assert_orig = [8, 4, 10, 3, 7]
    A = assert_orig[:]
    build_max_heap(A)
    assert A == [10, 7, 8, 3, 4], "BUILD-MAX-HEAP falló en el ejemplo"
    assert es_max_heap(A)

    # MAXIMUM no elimina
    assert maximum(A) == 10 and len(A) == 5

    # EXTRACT-MAX
    assert extract_max(A) == 10
    assert A == [8, 7, 4, 3], "EXTRACT-MAX falló en el ejemplo"
    assert es_max_heap(A)

    # Inserción: la prioridad 12 debe quedar en la raíz
    pos = insertar(A, 12)
    assert pos == 0 and A[0] == 12 and es_max_heap(A)

    # HeapSort del ejemplo
    assert heapsort([8, 4, 10, 3, 7])[1] == [3, 4, 7, 8, 10]

    # Casos borde
    assert heapsort([])[1] == []
    assert heapsort([5])[1] == [5]
    assert heapsort([2, 2, 2])[1] == [2, 2, 2]
    assert heapsort([5, 4, 3, 2, 1])[1] == [1, 2, 3, 4, 5]
    assert maximum([]) is None and extract_max([]) is None


    for _ in range(200):
        datos = [random.randint(-50, 50) for _ in range(random.randint(0, 30))]
        _, res = heapsort(datos)
        assert len(res) == len(datos)
        assert all(res[i] <= res[i + 1] for i in range(len(res) - 1))
        assert all(res.count(x) == datos.count(x) for x in set(datos))
        H = datos[:]
        build_max_heap(H)
        assert es_max_heap(H)
        extraidos = []
        while H:
            extraidos.append(extract_max(H))
            assert es_max_heap(H)
        assert all(extraidos[i] >= extraidos[i + 1] for i in range(len(extraidos) - 1))
    print("Todas las pruebas (asserts) pasaron correctamente.")

def main():
    ejemplo = [
        ("T01", "Compilar proyecto", 8),
        ("T02", "Generar backup", 4),
        ("T03", "Renderizar video", 10),
        ("T04", "Procesar correos", 3),
        ("T05", "Ejecutar pruebas", 7),
        ("T06", "Desplegar aplicación", 6),
    ]
    nueva_ejemplo = ("T07", "Actualizar servidor", 12)

    usar = input("¿Usar las 6 tareas de ejemplo? (s = ejemplo, n = ingresar manualmente): ").strip().lower()
    modo_ejemplo = usar == "s"

    print("\n" + "=" * 55 + "\nPUNTO 1. REGISTRO DE TAREAS\n" + "=" * 55)
    tareas = ejemplo[:] if modo_ejemplo else registrar_tareas()
    for t in tareas:
        print(f"  {t[0]} - {t[1]} - Prioridad {t[2]}")
    print("Lista original de prioridades:", prioridades(tareas))


    print("\n" + "=" * 55 + "\nPUNTO 2. CONSTRUCCIÓN DEL MAX-HEAP\n" + "=" * 55)
    heap = tareas[:]
    build_max_heap(heap)
    print("Lista original:", prioridades(tareas))
    print("Max-Heap:      ", prioridades(heap))
    print("Propiedad de Max-Heap válida:", es_max_heap(heap))
    print("\nÁrbol:")
    imprimir_arbol(heap)

    print("\n" + "=" * 55 + "\nPUNTO 3. COLA DE PRIORIDAD\n" + "=" * 55)
    print("A. MAXIMUM")
    m = maximum(heap)
    print(f"Tarea de mayor prioridad: {texto_tarea(m)}")
    print(f"Prioridad: {m[2]}")

    print("\nB. EXTRACT-MAX")
    procesada = extract_max(heap)
    print(f"Tarea procesada: {texto_tarea(procesada)}")
    print("Nuevo Heap:", prioridades(heap))
    imprimir_arbol(heap)

    print("\nC. INSERTAR NUEVA TAREA")
    if modo_ejemplo:
        nueva = nueva_ejemplo
        print(f"  Código: {nueva[0]}\n  Nombre: {nueva[1]}\n  Prioridad: {nueva[2]}")
    else:
        usados = {t[0] for t in tareas}
        nueva = leer_tarea(usados)
    pos = insertar(heap, nueva)
    print(f"Tarea insertada: {texto_tarea(nueva)} (prioridad {nueva[2]}) quedó en la posición {pos} del arreglo")
    print("Heap después de insertar:", prioridades(heap))
    print("Propiedad de Max-Heap válida:", es_max_heap(heap))
    imprimir_arbol(heap)

    print("\n" + "=" * 55 + "\nPUNTO 4. HEAPSORT\n" + "=" * 55)
    original = prioridades(tareas)
    max_heap, ordenada = heapsort(original)
    print("Lista original:", original)
    print("Max-Heap:      ", max_heap)
    print("Lista ordenada:", ordenada)


    print("\n" + "=" * 55 + "\nPUNTO 5. PRUEBAS\n" + "=" * 55)
    pruebas()

    print("\n" + "=" * 55 + "\nANÁLISIS DE COMPLEJIDAD (gráfica)\n" + "=" * 55)
    try:
        graficar_complejidad()
    except ImportError:
        print("matplotlib no está disponible; no se puede mostrar la gráfica.")


if __name__ == "__main__":
    main()
```
